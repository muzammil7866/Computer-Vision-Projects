```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# Load dataset
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize pixel values (0–255) to (0–1)
x_train, x_test = x_train / 255.0, x_test / 255.0

# Visualize some images
plt.figure(figsize=(5,5))
for i in range(9):
    plt.subplot(3,3,i+1)
    plt.imshow(x_train[i], cmap='gray')
    plt.title(f"Label: {y_train[i]}")
    plt.axis('off')
plt.tight_layout()
plt.show()

# Build a basic neural network model
model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),        # Converts image to 1D array
    layers.Dense(128, activation='relu'),        # Hidden layer
    layers.Dense(10, activation='softmax')       # Output layer (10 classes: digits 0–9)
])

# Compile model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train model
model.fit(x_train, y_train, epochs=5)

# Evaluate on test data
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"Test Accuracy: {test_acc:.2f}")

# Predict on a test image
import numpy as np
image = x_test[0]
plt.imshow(image, cmap='gray')
plt.title("Sample Test Image")
plt.axis('off')
plt.show()

prediction = model.predict(np.expand_dims(image, 0))
predicted_label = np.argmax(prediction)
print(f"Model Prediction: {predicted_label}")
```

```python
import tensorflow as tf
from tensorflow.keras import models, layers
import matplotlib.pyplot as plt

#now we are loading the dataset

mnist = tf.keras.datasets.mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data() #train first then test

#now we are normalising the pixel values

x_train, x_test = x_train/255.0, x_test/255.0


#now we are about to visualise some of the images to have an idea of how the dataset is loaded

plt.figure(figsize = (5,5))

for i in range(9):
    plt.subplot(3,3,i+1)
    plt.imshow(x_train[i], cmap = 'gray')
    plt.title(f"Label: {y_train[i]}")
    plt.axis('on')

plt.tight_layout()
plt.show()

# Now we will build a basic neural network model

model = models.Sequential([

    # converts the image to 1d array
    layers.Flatten(input_shape = (28,28)),

    #hidden layer of 128 neurons and activation function as relu
    layers.Dense(128, activation = "relu"),

    #output layer of 10 neurons (w.r.t. 10 target classes) and activation function of softmax
    layers.Dense(10, activation = 'softmax')
])


# now we will compile the model

model.compile(optimizer = 'adam', loss = 'sparse_categorical_crossentropy', metrics = ['accuracy'])

model.fit(x_train, y_train, epochs = 5)

test_loss, test_acc = model.evaluate(x_test, y_test)

print("Test Accuracy is: ", test_acc)


import numpy as np

img = x_test[0]
plt.imshow(img, cmap = 'gray')
plt.title(f"Label: {y_test[0]}")
plt.axis('off')

plt.show()

prediction = model.predict(np.expand_dims(img,0))
print(prediction)
prediction_label = np.argmax(prediction)
print("Predicted Label: ", prediction_label)






```
