```python
import os
import shutil
import glob
import xml.etree.ElementTree as ET
import pandas as pd
import numpy as np
import cv2
import yaml
import torch
import random
import warnings
from sklearn.model_selection import train_test_split
from ultralytics import YOLO
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt

# Filter warnings
warnings.filterwarnings('ignore')

print(f"CUDA Available: {torch.cuda.is_available()}")

# ==========================================
# --- CONFIGURATION ---
# ==========================================
INPUT_ROOT = '/kaggle/input' 
WORKING_DIR = '/kaggle/working'
DATASET_NAME = 'FracAtlas_YOLO'
IMG_SIZE = 512
SEED = 42

# Training Requirements
EPOCHS = 15
BATCH_SIZE = 16
LR = 0.0001
OPTIMIZER = 'Adam'
PROJECT_NAME = 'FracAtlas_Project'
RUN_NAME = 'YOLOv11n_Run'

# ==========================================
# 1. ROBUST DATASET FINDER
# ==========================================
# We look for the 'FracAtlas' folder specifically
dataset_root = None
for root, dirs, files in os.walk(INPUT_ROOT):
    if 'images' in dirs and 'Annotations' in dirs:
        dataset_root = root
        break

if dataset_root is None:
    dataset_root = '/kaggle/input/fracatlas-original-dataset/FracAtlas'

print(f"Dataset root found at: {dataset_root}")

images_dir = os.path.join(dataset_root, 'images')
# Explicitly target PASCAL VOC folder seen in your screenshots
xml_dir = os.path.join(dataset_root, 'Annotations', 'PASCAL VOC')

# Fallback if PASCAL VOC folder is named differently or missing
if not os.path.exists(xml_dir):
    xml_dir = os.path.join(dataset_root, 'Annotations')

print(f"Searching for images in: {images_dir}")
print(f"Searching for labels in: {xml_dir}")

# ==========================================
# 2. SMART XML MAPPING
# ==========================================
print("Building XML Map...")
xml_map = {}
for root, dirs, files in os.walk(xml_dir):
    for file in files:
        if file.lower().endswith('.xml'):
            base_name = os.path.splitext(file)[0]
            xml_map[base_name] = os.path.join(root, file)

print(f"Found {len(xml_map)} XML files.")

# ==========================================
# 3. GATHER & MATCH DATA
# ==========================================
print("Gathering Images...")
all_image_paths = []
all_xml_paths = []

# Scan images recursively
image_files = []
for ext in ('*.jpg', '*.jpeg', '*.png', '*.JPG'):
    image_files.extend(glob.glob(os.path.join(images_dir, '**', ext), recursive=True))

print(f"Total Images Scanned: {len(image_files)}")

matched_count = 0
for img_path in image_files:
    filename = os.path.basename(img_path)
    base_name = os.path.splitext(filename)[0]
    
    # Try to find corresponding XML
    xml_path = xml_map.get(base_name)
    
    all_image_paths.append(img_path)
    all_xml_paths.append(xml_path)
    
    if xml_path:
        matched_count += 1

print(f"Matched {matched_count} images with XML labels.")

if matched_count == 0:
    raise ValueError("STOP: No XML labels matched! Check dataset structure.")

# ==========================================
# 4. XML PARSER
# ==========================================
def convert_xml_to_yolo(xml_file, img_width, img_height):
    try:
        tree = ET.parse(xml_file)
        root = tree.getroot()
        yolo_labels = []
        
        for obj in root.findall('object'):
            class_id = 0 
            
            bndbox = obj.find('bndbox')
            if bndbox is None: continue
                
            xmin = float(bndbox.find('xmin').text)
            ymin = float(bndbox.find('ymin').text)
            xmax = float(bndbox.find('xmax').text)
            ymax = float(bndbox.find('ymax').text)
            
            if xmax <= xmin or ymax <= ymin: continue

            x_center = ((xmin + xmax) / 2) / img_width
            y_center = ((ymin + ymax) / 2) / img_height
            width = (xmax - xmin) / img_width
            height = (ymax - ymin) / img_height
            
            x_center = max(0, min(1, x_center))
            y_center = max(0, min(1, y_center))
            width = max(0, min(1, width))
            height = max(0, min(1, height))
            
            yolo_labels.append(f"{class_id} {x_center} {y_center} {width} {height}")
            
        return yolo_labels
    except:
        return []

# ==========================================
# 5. SPLIT & PROCESS
# ==========================================
if os.path.exists(os.path.join(WORKING_DIR, 'datasets')):
    shutil.rmtree(os.path.join(WORKING_DIR, 'datasets'))

train_imgs, temp_imgs, train_xmls, temp_xmls = train_test_split(all_image_paths, all_xml_paths, test_size=0.3, random_state=SEED)
val_imgs, test_imgs, val_xmls, test_xmls = train_test_split(temp_imgs, temp_xmls, test_size=0.66, random_state=SEED)

base_out = os.path.join(WORKING_DIR, 'datasets', DATASET_NAME)
for split in ['train', 'val', 'test']:
    os.makedirs(os.path.join(base_out, 'images', split), exist_ok=True)
    os.makedirs(os.path.join(base_out, 'labels', split), exist_ok=True)

total_labels_written = 0

def process_batch(img_list, xml_list, split_name):
    global total_labels_written
    for img_path, xml_path in tqdm(zip(img_list, xml_list), total=len(img_list), desc=f"Processing {split_name}"):
        filename = os.path.basename(img_path)
        shutil.copy(img_path, os.path.join(base_out, 'images', split_name, filename))
        
        dst_label = os.path.join(base_out, 'labels', split_name, os.path.splitext(filename)[0] + '.txt')
        
        if xml_path:
            img = cv2.imread(img_path)
            if img is not None:
                h, w, _ = img.shape
                labels = convert_xml_to_yolo(xml_path, w, h)
                if labels:
                    with open(dst_label, 'w') as f:
                        f.write('\n'.join(labels))
                    total_labels_written += 1
                else:
                    open(dst_label, 'w').close()
        else:
            open(dst_label, 'w').close()

process_batch(train_imgs, train_xmls, 'train')
process_batch(val_imgs, val_xmls, 'val')
process_batch(test_imgs, test_xmls, 'test')

# ==========================================
# 6. TRAINING
# ==========================================
if total_labels_written > 0:
    yaml_content = {
        'path': base_out,
        'train': 'images/train',
        'val': 'images/val',
        'test': 'images/test',
        'names': {0: 'fracture'}
    }
    yaml_path = os.path.join(WORKING_DIR, 'fracatlas.yaml')
    with open(yaml_path, 'w') as f:
        yaml.dump(yaml_content, f)

    print("\nStarting Training...")
    # Clean previous project runs to avoid confusion if possible, 
    # OR we use the dynamic plotter below.
    model = YOLO('yolo11n.pt') 

    model.train(
        data=yaml_path,
        epochs=EPOCHS,
        imgsz=IMG_SIZE,
        batch=BATCH_SIZE,
        optimizer=OPTIMIZER,
        lr0=LR,
        degrees=5.0, fliplr=0.5, hsv_v=0.4, scale=0.5,
        project=PROJECT_NAME,
        name=RUN_NAME,
        verbose=True
    )
else:
    print("CRITICAL: Training skipped. No labels generated.")

# ==========================================
# 7. DYNAMIC PLOTTING (THE FIX)
# ==========================================
if total_labels_written > 0:
    print("\n--- Plotting Results ---")
    
    # Find the LATEST run folder (e.g., YOLOv11n_Run, YOLOv11n_Run2, etc.)
    # This fixes the issue of plotting empty graphs from failed runs.
    possible_runs = glob.glob(os.path.join(PROJECT_NAME, f"{RUN_NAME}*"))
    if possible_runs:
        # Sort by modification time to get the newest one
        latest_run = max(possible_runs, key=os.path.getmtime)
        print(f"Visualizing results from: {latest_run}")
        
        results_path = os.path.join(latest_run, 'results.csv')
        
        if os.path.exists(results_path):
            df = pd.read_csv(results_path)
            df.columns = [x.strip() for x in df.columns]

            plt.figure(figsize=(12, 5))
            
            # Box Loss
            plt.subplot(1, 2, 1)
            plt.plot(df['train/box_loss'], label='Train Box')
            plt.plot(df['val/box_loss'], label='Val Box')
            plt.title('Box Loss')
            plt.xlabel('Epochs')
            plt.legend()
            
            # Class Loss
            plt.subplot(1, 2, 2)
            plt.plot(df['train/cls_loss'], label='Train Class')
            plt.plot(df['val/cls_loss'], label='Val Class')
            plt.title('Class Loss')
            plt.xlabel('Epochs')
            plt.legend()
            
            plt.show()
            
            # Save Best Model
            best_pt = os.path.join(latest_run, 'weights', 'best.pt')
            if os.path.exists(best_pt):
                shutil.copy(best_pt, 'FracAtlas_YOLOv11n.pt')
                print("✅ Model saved as FracAtlas_YOLOv11n.pt")
        else:
            print(f"Warning: results.csv not found in {latest_run}")
    else:
        print("Error: No training runs found.")
```
